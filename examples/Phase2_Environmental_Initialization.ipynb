````python
import json
from pathlib import Path
root = Path.cwd()  # Run from your SCE repository root
manifest_path = root / 'phase1_output/trials_manifest.json'
manifest = json.loads(manifest_path.read_text())
for trial in manifest['trials']:
    print(trial['trial_id'],trial['site'],trial['planting_date'],trial['maturity_class'] or 'MG requires resolution')
````
