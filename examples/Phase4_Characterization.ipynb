# SCE v2 Phase 4 — Stage environmental characterization
Run after the completed native Phase 3 APSIM simulation. This reads the SQLite Report and produces daily, stage and tidy covariate files.

```python
from pathlib import Path
import subprocess
ROOT = Path.cwd()
DB = ROOT / 'phase3_native_test/SCE_Phase3_2001_mid000.db'
OUT = ROOT / 'phase4_output_2001_mid000'
assert DB.exists(), f'Missing Phase 3 database: {DB}'
if OUT.exists() and (OUT/'stage_summary.csv').exists():
    OUT = ROOT/'phase4_output_2001_mid000_second_run'
cmd=['python3',str(ROOT/'scripts/characterize_season.py'),'--db',str(DB),'--output-dir',str(OUT)]
print(' '.join(cmd))
r=subprocess.run(cmd,capture_output=True,text=True)
print(r.stdout or r.stderr)
assert r.returncode == 0
```

```python
import pandas as pd
s = pd.read_csv(OUT/'stage_summary.csv')
d = pd.read_csv(OUT/'daily_phenology.csv')
c = pd.read_csv(OUT/'stage_covariates.csv')
display(s[['stage_segment_id','stage_name','days','Rain_sum','Radn_sum','ThermalTime_sum','WaterStress_mean']].head(20))
print('daily rows',len(d),'stage segments',len(s),'tidy covariates',len(c))
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(12,4))
ax.plot(pd.to_datetime(d['Date']),d['Rain'].fillna(0).cumsum(),label='Cumulative rainfall (mm)')
ax.set_xlabel('Date');ax.set_ylabel('Cumulative precipitation (mm)');ax.legend();plt.show()
```
